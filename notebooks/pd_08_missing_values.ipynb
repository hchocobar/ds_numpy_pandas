## Valores faltantes

Ya vimos un adelanto de cómo Pandas maneja los valores faltantes usando el tipo `None` y los valores `NaN` de NumPy. Los valores faltantes son bastante comunes en las actividades de limpieza de datos. Pueden deberse a diversas razones, y aquí solo mencionaré algunas.

Por ejemplo, si realizas una encuesta y un participante no contesta una pregunta, el valor faltante es en realidad una omisión. Este tipo de datos faltantes se denomina:

- **Missing at Random** si existen otras variables que podrían usarse para predecir la variable faltante.
- Si no existe relación con otras variables, entonces denominamos a estos datos **Missing Completely at Random (MCAR)**.

Estos son solo dos ejemplos de datos faltantes, y existen muchos más. Por ejemplo, puede que falten datos porque no se recopilaron, ya sea por el proceso responsable de su recopilación, como un investigador, o porque no tendría sentido recopilarlos. Este último ejemplo es muy común al combinar DataFrames de múltiples fuentes, como al combinar una lista de personas de una universidad con una lista de oficinas de la misma (los estudiantes generalmente no tienen oficinas).

Veamos algunas maneras de manejar los datos faltantes en pandas.

```python
import pandas as pd
```

Pandas es bastante bueno detectando valores faltantes directamente de los formatos de datos subyacentes, como los archivos CSV.

Aunque la mayoría de los valores faltantes suelen tener el formato NaN, NULL, None o N/A, a veces no están etiquetados con tanta claridad.

Por ejemplo, que utilicen el valor 99 en categorías binarias para indicar un valor faltante. La función `read_csv()` de pandas tiene un parámetro llamado `na_values` que nos permite especificar el formato de los valores faltantes. Permite usar escalares, cadenas, listas o diccionarios.

### Ejemplo 1

```python
df = pd.read_csv('../data/class_grades.csv')
df.head(10)
```

Podemos usar la función `.isnull()` para crear una máscara booleana de todo el dataframe.

```python
mask = df.isnull()
mask.head(10)
```

Esto puede ser útil para procesar filas según ciertas columnas de datos.

Otra operación útil es poder eliminar todas las filas que contengan datos faltantes, lo cual se puede hacer con la función `dropna()`.

```python
df.dropna().head(10)
```

> Nota: Observa cómo las filas con índices 2, 3, 7 y 11 ya no están.

Una de las funciones útiles de Pandas para trabajar con valores faltantes es la función de relleno, `fillna()`. Esta función acepta un número o parámetros.

Podrías pasar un único valor, llamado escalar, para reemplazar todos los datos faltantes con un solo valor. Si bien esto no aplica en este caso, es un uso bastante común.

Por lo tanto, si quisiéramos rellenar todos los valores faltantes con 0, usaríamos `fillna`.

```python
df.fillna(0, inplace=True)
df.head(10)
```

> Tenga en cuenta que el atributo inplace hace que pandas complete los valores en línea y no devuelve una copia del dataframe, sino que modifica el dataframe que tiene.

Además de las reglas que controlan cómo se cargan los valores faltantes, a veces es útil considerar que estos valores contienen información.

### Ejemplo 2

```python
df = pd.read_csv("../data/log.csv")
df.head(20)
```

En estos datos, la primera columna es una marca de tiempo en formato Unix. La siguiente columna contiene el nombre de usuario, seguido de la página web que visita y el vídeo que reproduce. Cada fila del DataFrame indica la posición de reproducción.

Podemos observar que, a medida que la posición de reproducción aumenta en uno, la marca de tiempo aumenta aproximadamente 30 segundos.

Excepto para el usuario Bob! Resulta que Bob ha pausado su reproducción, por lo que, a medida que avanza el tiempo, la posición de reproducción no cambia.

Nótese también la dificultad de extraer esta información de los datos, ya que no están ordenados por marca de tiempo como cabría esperar. Esto es bastante común en sistemas con un alto grado de paralelismo. Hay muchos valores faltantes en las columnas de pausa y volumen. No es eficiente enviar esta información a través de la red si no ha cambiado. Por lo tanto, este sistema en particular simplemente inserta valores nulos en la base de datos si no hay cambios.

En Pandas, podemos ordenar por índice o por valores. Aquí, simplemente convertiremos la marca de tiempo en un índice y luego ordenaremos por índice.

```python
df = df.set_index('time')
df = df.sort_index()
df.head(20)
```

Si analizamos detenidamente la salida, observaremos que el índice no es realmente único. Dos usuarios parecen poder usar el sistema simultáneamente. De nuevo, un caso muy común.

Restablezcamos el índice y, en su lugar, usemos indexación multinivel por tiempo y usuario, promoviendo el nombre de usuario a un segundo nivel del índice para solucionar este problema.

```python
df = df.reset_index()
df = df.set_index(['time', 'user'])
df
```

Ahora que tenemos los datos indexados y ordenados correctamente, podemos completar los datos faltantes usando `ffill`.

```python
df = df.fillna(method='ffill')
df.head()
```

### Ejemplo 3

También podemos personalizar el relleno para reemplazar valores con la función `replace()`. Permite el reemplazo mediante varios métodos: valor a valor, lista, diccionario y expresiones regulares.

Generemos un ejemplo sencillo.

```python
df = pd.DataFrame({'A': [1, 1, 2, 3, 4],
                   'B': [3, 6, 3, 8, 9],
                   'C': ['a', 'b', 'c', 'd', 'e']})
df
```

```python
# Reemplazamos valor por valor
df.replace(1, 100)
```

```python
# Reemplazamos valores por lista
df.replace([1, 3], [100, 300])
```

### Ejemplo 4

La función de reemplazo de pandas es que también admite expresiones regulares.


```python
df = pd.read_csv("../data/log.csv")
df.head(20)
```

Para reemplazar usando una expresión regular, definimos:
- el primer parámetro como el patrón que queremos buscar,
- el segundo como el valor que queremos mostrar al encontrar una coincidencia y,
- finalmente, pasamos un tercer parámetro: `regex=True`.

Imagina que queremos detectar todas las páginas HTML en la columna "video" (es decir, las que terminan en ".html") y reemplazarlas con la palabra clave "webpage".

```python
df.replace(to_replace=".*.html$",
           value="webpage",
           regex=True)
```

> Nota:
>
> Una última nota sobre los valores faltantes. Al usar funciones estadísticas en DataFrames, estas funciones suelen ignorar los valores faltantes.
>
> Por ejemplo, si intentas calcular la media de un DataFrame, la función de NumPy ignorará los valores faltantes. Esto suele ser lo que se busca, pero debes tener en cuenta que se están excluyendo valores. El motivo de la presencia de valores faltantes depende del problema que se esté intentando resolver.
