## Cargando e indexando un DataFrame

En esta lección, demostraré cómo cargar datos desde un archivo CSV a un DataFrame.

Cualquier programa de hojas de cálculo pueden guardar o exportar sus resultados en formato CSV (archivo de valores separados por comas). Es un formato bastante flexible, increíblemente ligero y totalmente omnipresente.

### Ejemplo 1

```python
df = pd.read_csv('../data/admission-predict.csv')

df.head()
```

Observamos que, por defecto, el índice comienza en 0, mientras que el número de serie de los estudiantes comienza en 1. Por lo tanto, vemos que Pandas ha creado un nuevo índice.

Podemos establecer que la columna `Serial No.` se establezca como índice utilizando la sentencia `index_col`.

```python
df = pd.read_csv('../data/admission-predict.csv', index_col=0)

df.head()
```

```python
# Podemos usar el atributo columns del dataframe para obtener una lista.
df.columns
```

Observaciones:

- Las columnas "LOR " y "Chance of Admit." tienen espacios en blanco al final.
- Tenemos dos columnas, "SOP" y "LOR " que probablemente no todos sepan qué significan.

1. Para eliminar los espacios en blanco podemos crear una función que realice la limpieza y luego indicarle a `rename` que la aplique a todos los datos. Python incluye una práctica función de cadena llamada `strip()` para eliminar espacios en blanco. Al pasar esta función a `rename`, la pasamos como parámetro del `mapper` y luego indicamos si el eje debe ser columnas o índice

```python
# Eliminamos los espacios en blanco al final del nombre de las columnas
df = df.rename(mapper=str.strip, axis='columns')

df.columns
```

2. Ahora vamos a cambiar los nombres de nuestras columnas para que sea más claro. En Pandas, podemos usar la función `rename()`. Esta función recibe un parámetro llamado `columns`, al cual debemos pasarle un diccionario donde las claves sean los nombres de las columnas antiguas y los valores sean los nombres de las nuevas columnas nuevas.

```python
df = df.rename(columns={'SOP': 'Statement of Purpose',
                        'LOR': 'Letter of Recommendation'})
df.columns
```

```python
df.head()
```

### Ejemplo 2

También podemos usar el atributo `df.columns` asignándole una lista de nombres de columna, lo que renombrará directamente las columnas.

Esto es muy eficiente, especialmente cuando se tienen muchas columnas y solo se desea cambiar unas pocas. Esta técnica tampoco se ve afectada por errores sutiles en los nombres de las columnas, un problema que acabamos de encontrar. Con una lista, se puede usar el índice para cambiar un valor específico o usar la comprensión de lista para cambiar todos los valores.

```python
# Cargamos nuevamente el archivo original
df = pd.read_csv('../data/admission-predict.csv')

df.head()
```

```python
df.columns
```

```python
# Obtenemos una lista de las columnas
cols = list(df.columns)

# Utilizamos list comprehenshion para cambiar los nombres de las columnas
cols = [col.lower().strip().replace(' ', '_') for col in cols]
df.columns = cols

df.columns
```
