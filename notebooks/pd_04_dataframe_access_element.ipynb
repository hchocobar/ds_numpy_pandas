## ¿Cómo accedemos a los elementos de un DataFrame?

```python
import pandas as pd
```

La estructura de datos DataFrame es fundamental en la biblioteca Pandas. Es el objeto principal con el que trabajarás en tareas de análisis y limpieza de datos.

Conceptualmente, un DataFrame es un objeto de serie bidimensional, con un índice y varias columnas de contenido, cada una con una etiqueta.

```python
students = [{'Name': 'Alice', 'Class': 'Physics', 'Score': 85},
            {'Name': 'Jack', 'Class': 'Chemistry', 'Score': 82},
            {'Name': 'Helen', 'Class': 'Biology', 'Score': 90}]

df = pd.DataFrame(students, index=['school1', 'school2', 'school1'])

df.head()
```

### Accediendo mediante `.loc` y `loc`

Al igual que con las series, podemos extraer datos usando los atributos `.iloc` y `.loc`. Dado que el DataFrame es bidimensional, pasar un solo valor al operador de indexación `loc` devolverá la serie si solo hay una fila.

```python
df.loc['school2']
```

```python
# Comprobamos el type() devuelto
type(df.loc['school2'])
```

Es importante recordar que los índices y los nombres de las columnas, tanto en el eje horizontal como en el vertical, podrían no ser únicos. En este ejemplo, vemos dos registros para la escuela1 en filas diferentes.

Si usamos un solo valor con el atributo `loc` del DataFrame, se devolverán varias filas del DataFrame, no como una nueva serie, sino como un nuevo DataFrame.

```python
df.loc['school1']
```

```python
# Comprobamos el type() devuelto
type(df.loc['school1'])
```

Una de las ventajas del DataFrame de Pandas es que permite seleccionar datos rápidamente según varios ejes.

Por ejemplo, si solo se quisiera listar los nombres de los estudiantes de la `school1`, se proporcionarían dos parámetros a `.loc`: el índice de la fila y el nombre de la columna.

```python
df.loc['school1', 'Name']
```

Recuerdemos que, al igual que con las Series, los desarrolladores de pandas implementaron esto usando el operador de indexación y no como parámetros de una función.

¿Qué haríamos si quisiéramos seleccionar una sola columna? Existen varios mecanismos. Primero, podríamos transponer la matriz. Esto convierte todas las filas en columnas y viceversa, y se realiza con el atributo `T`.

```python
df.T
```

Entonces podemos llamar a `.loc` de la transpuesta para obtener los nombres de los estudiantes

```python
df.T.loc['Name']
```

Sin embargo, dado que `iloc` y `loc` se utilizan para la selección de filas, Pandas reserva el operador de indexación directamente en el DataFrame para la selección de columnas. En un DataFrame de Pandas, las columnas siempre tienen un nombre.

Por lo tanto, esta selección siempre se basa en etiquetas y no resulta tan confusa

```python
df['Name']
```

```python
type(df['Name'])
```

Dado que el resultado de usar el operador de indexación es un DataFrame o una Serie, se pueden encadenar operaciones.

Por ejemplo, podemos seleccionar todas las filas relacionadas con la `school1` usando `.loc`, y luego proyectar la columna `name` solo a partir de esas filas.

```python
df.loc['school1']['Name']
```

```python
# Verificamos el type() devuelto
print(type(df.loc['school1']))  # DataFrame
print(type(df.loc['school1']['Name']))  # Series
```

El atributo `.loc` selecciona filas y puede recibir dos parámetros:
- el índice de la fila y
- la lista de nombres de columna.

El atributo `.loc` también admite el rebanado.

Si quisiéramos seleccionar todas las filas, podemos utilizar dos puntos para indicar un rebanado completo de principio a fin.

Esto es similar a rebanar caracteres en una lista en Python. Luego, podemos agregar el nombre de la columna como segundo parámetro, en forma de cadena. Si quisiéramos incluir varias columnas, podríamos hacerlo en una lista. Pandas devolverá únicamente las columnas solicitadas.

Aquí tienes un ejemplo donde solicitamos los nombres y puntuaciones de todas las escuelas usando el operador `.loc`.

```python
df.loc[:,['Name', 'Score']]
```

> Nota: Los dos puntos indican que queremos obtener todas las filas, y la lista en la segunda posición del argumento es la lista de columnas que queremos recuperar.

### Conclusiones:

El DataFrame se indexa tanto por filas como por columnas, y puedes seleccionar fácilmente filas individuales y proyectar las columnas que te interesen usando los métodos de indexación de la clase Series.
